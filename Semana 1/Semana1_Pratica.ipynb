# Projeto Anáalise e Desnvolvimento de Dados

### Passo 1: Simulação de Entrada de Dados

```python
# Simulando a "Tabela" do Banco de Dados

estoque = [
{"id": 1, "nome": "Notebook", "cat": "TI", "preco": 3500.0, "qtd": 10},
{"id": 2, "nome": "Mouse", "cat" : "TI", "preco": 120.0, "qtd": 50},
{"id": 3, "nome": "Teclado", "cat" : "TI", "preco": 100.0, "qtd": 60},
{"id": 4, "nome": "Monitor", "cat" : "TI", "preco": 950.0, "qtd": 10},
{"id": 5, "nome": "Cadeira", "cat" : "Moveis", "preco": 1200.0, "qtd": 8},
{"id": 6, "nome": "Mesa", "cat" : "Moveis", "preco": 800.0, "qtd": 5},
{"id": 7, "nome": "Headset", "cat" : "TI", "preco": 420.0, "qtd": 20},
 ]
```

### Passo 2: Tansformação Analítica

```python
import pandas as pd

df = pd.DataFrame (estoque)

# Calculando o Valor Total em Estoque por Produto
df['valor_total'] = df['preco'] * df['qtd']

# Insight 1: Qual o valor total investido em estoque?
total_geral = df['valor_total'].sum()

# Insigth 2: Qual a media de preco por categoria?
media_cat = df.groupby('cat')['preco'].mean()

# Insight 3: Qual o valor de investimento por categoria?
total_categoria = df.groupby('cat')['valor_total'].sum()


print (f"Investimento Total: R$ {total_geral}")
print (f"O total de investimento por caregoria foi: {total_categoria}")
print(media_cat)
```

O código cria uma lista de dados simples, de uma loja de venda de itens de informática, a loja possui 2 categorias de itens, sendo elas Móveis e Equipamentos Eletronicos classificada como "TI", e faz calculos simples referentes a estoque e produtos.

Foi utilizada a biblioteca Pandas para converter a lista de produtos em Dataframe e fazer os calculos simples.
